```python
players = [
    {"Player Name": "Serena Williams", "Sport": "Tennis", "Achievements": 23},
    {"Player Name": "Lionel Messi", "Sport": "Soccer", "Achievements": 7},
    {"Player Name": "Michael Phelps", "Sport": "Swimming", "Achievements": 23},
    {"Player Name": "Usain Bolt", "Sport": "Athletics", "Achievements": 8},
    {"Player Name": "Roger Federer", "Sport": "Tennis", "Achievements": 20},
    {"Player Name": "Cristiano Ronaldo", "Sport": "Soccer", "Achievements": 5}
]
```

```python
for p in players:
    if p["Player Name"] == "Lionel Messi":
        if p["Achievements"] > 10:
            print(f"Name: {p['Player Name']}, Sport: {p['Sport']}, Achievements: {p['Achievements']}")
        else:
            print("does not have more than 10 achievements.")
```

```python
for p in players:
    if p["Sport"] == "Tennis" or p["Achievements"] == 20:
        print(f"Success: {p['Player Name']} matches the criteria.")
```

```python
num_str = input("Enter an integer between 1 and 9999: ")
num = int(num_str)
if num % 2 == 0:
    print(f"{num} is Even.")
else:
    print(f"{num} is Odd.")
if num_str == num_str[::-1]:
    print(f"{num} is a Palindrome.")
else:
    print(f"{num} is not a Palindrome.")
digit_sum = sum(int(digit) for digit in num_str)
if num % digit_sum == 0:
    print(f"{num} is divisible by the sum of its digits ({digit_sum}).")
else:
    print(f"{num} is not divisible by the sum of its digits ({digit_sum}).")
```

```python
squares = ['orange', 'orange', 'green', 'purple', 'blue'] 
new_squares = []
i = 0

while i < len(squares) and squares[i] == 'orange':
    new_squares.append(squares[i])
    i += 1

print("New list:", new_squares)
```

```python
animals = ["lion", "giraffe", "gorilla", "parrots", "crocodile", "deer", "swan"]
seven_letter_animals = []
i = 0

while i < len(animals):
    if len(animals[i]) == 7:
        seven_letter_animals.append(animals[i])
    i += 1

print("Animals with 7 letters:", seven_letter_animals)
```
